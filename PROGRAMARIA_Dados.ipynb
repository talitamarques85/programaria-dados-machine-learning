# 5- Identificação e Tratamento de Outliers
Nesta etapa, utilizamos a biblioteca **Matplotlib**  para a construção de gráficos (como *Boxplots*) que permitem visualizar e identificar a presença de valores discrepantes (outliers) na distribuição dos dados.
* **Objetivo:** Analisar o impacto desses valores extremos e aplicar as devidas correções (remoção ou limite) para não distorcer as análises estatísticas.

```python
#ferramenta de visualização
import matplotlib.pyplot as plt
```

```python
#visualizar outliers em salario
plt.boxplot(dados['SALARIO'])
```

```python
#Aqui encontramos os valores medio do primeiro e terceiro quartil
#uma forma de identificar e tratar outliers
Q1 = dados['SALARIO'].quantile(0.25)
print(Q1)
Q3 = dados['SALARIO'].quantile(0.75)
print(Q3)
```

```python
#estamos em busca de encontrar os valores limites mostrado do boxplot
#para isso encontramos o valor do interquartil que e o terceiro quartil - o primeiro quartil
IQ = Q3 - Q1
IQ
```

```python
#Agora sim calculamos os limites - primeiro o superior que e dado pelo seguinte calculo
lim_superior = Q3 + (1.5*IQ)
print(lim_superior)
#O limite inferior é dado pelo calculo:
lim_inferior = Q1 - (1.5*IQ)
print(lim_inferior)
```

```python
#Aqui vamos analisar a faixa salarial para ter mais informacoes dos dados
dados['FAIXA SALARIAL'].value_counts()
#nesse contexto vemos que existem valores acima do limite superior que devem ser considerados
#e e possivel tratar esses valores atraves do calculo do desvio padra.
```

```python
#primeiro calculamos a media da coluna salario
media_salario = dados['SALARIO'].mean()
print(media_salario)
```

```python
#Agora o calculo do desvio padra
desviop_salario = dados['SALARIO'].std()
print(desviop_salario)
```

```python
#para tratar esses valores discrepantes podemos usar media + X*desvio padrao
#esse calculo depende de analise de contexto
#no nosso contexto os outliers estao muito distantes da media
#decisao de negocio e usar media +3*desvio padrao
limite_superior = media_salario + (3*desviop_salario)
print(limite_superior)
#Aqui definimos que todos os valores acima de 64.800 sao outliers
```

# 7 - Análise de Associação e Correlação Estatística

Nesta etapa, exploramos as relações entre as variáveis da base para identificar dependências e selecionar as melhores *features* para a modelagem. Como o conjunto de dados possui variáveis numéricas e categóricas, aplicamos métodos estatísticos distintos para cada tipo de dado:

1. **Variáveis Numéricas Contínuas:** Coeficiente de Correlação de Pearson + Validação Hipotética via `SciPy`.
2. **Variáveis Categóricas:** Coeficiente $V$ de Cramér (baseado no teste Qui-Quadrado / $\chi^2$).

```python
#aqui vamos usar uma funcao do pandas para determinar correlacao
#0.7 a 1.0 (ou -0.7 a -1.0): Correlação muito forte.
#0.5 a 0.7 (ou -0.5 a -0.7): Correlação forte.
#0.3 a 0.5 (ou -0.3 a -0.5): Correlação moderada.
#0.1 a 0.3 (ou -0.1 a -0.3): Correlação fraca.
#0.0 a 0.1 (ou -0.0 a -0.1): Correlação muito fraca ou insignificante
#aqui para correlacao continua, mais comuns para dados numericos
#chamada coeficente de Person

correlacao_continua = dados['IDADE'].corr(dados['SALARIO'])
correlacao_continua

#A correlacao é moderada, portanto quanto maior idade nao determina um melhor salario
```

```python
#essa e a funcao do quiquadrado
#importada do scipy
from scipy.stats import chi2_contingency
```

```python
def cramer_coeficiente(coluna1, coluna2):
  tab_cruzada = np.array(pd.crosstab(coluna1, coluna2))
  chi2 = chi2_contingency(tab_cruzada)[0]
  soma = np.sum(tab_cruzada)
  mini = min(tab_cruzada.shape)-1
  cramer = np.sqrt(chi2/(soma*mini))
  return cramer
```

```python
cramer_coeficiente(dados['COR/RACA/ETNIA'], dados['NIVEL DE ENSINO'])
```

```python
tab_cruzada = pd.crosstab(dados['COR/RACA/ETNIA'], dados['NIVEL DE ENSINO'])
tab_cruzada
```

# 9 -  Análise Relacional e SQL via DBeaver (SQLite)

Cruzamento da base principal com dados demográficos municipais (status_brasil) utilizando **SQLite** e **DBeaver**.

* **Manipulação (DDL/DML):** Ajustes de colunas (ALTER TABLE) e atualização de dados (UPDATE).
* **Cruzamento de Dados:** Unificação de tabelas via INNER JOIN (chave Municipio_ID).
* **Análise Demográfica:** Agrupamentos por estado e região (GROUP BY, SUM`, COUNT) para avaliar população e representatividade.

```python
import sqlite3
```

```python
conexao = sqlite3.connect('/content/status_brasil')
```

```python
dados = pd.read_csv('/content/drive/MyDrive/PYTHON/analisedados.csv')
```

```python
lista_estados = list(dados['UF ONDE MORA'].dropna().unique())
lista_estados
```

```python
query = '''SELECT Municipios_Brasileiros.Estado, AVG(Municipio_Status.renda)
    FROM Municipios_Brasileiros
    INNER JOIN Municipio_Status ON Municipios_Brasileiros.Municipio_ID = Municipio_Status.municipio_ID
    WHERE Municipios_Brasileiros.Estado IN ({})
    GROUP BY Municipios_Brasileiros.Estado;'''.format(','.join(['?' for _ in lista_estados]))
```

```python
estado_renda = pd.read_sql(query, con=conexao, params=lista_estados)
```

```python
estado_renda.rename(columns={'Estado': 'ESTADO', 'AVG(Municipio_Status.renda)': 'Renda_Media_Estado'}, inplace=True)
dados.rename(columns={'UF ONDE MORA': 'ESTADO'}, inplace=True)
dados = dados.merge(estado_renda, on='ESTADO', how='left')
```

```python
correlacao_renda_salario = dados['SALARIO'].corr(dados['Renda_Media_Estado'])
display(correlacao_renda_salario)
```
